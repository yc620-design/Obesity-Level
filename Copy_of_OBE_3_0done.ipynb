```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Encode the target variable: Convert 'NObeyesdad' (Obesity Level) to numeric labels (0, 1, 2, 3...)
le = LabelEncoder()
obe_cleaned['Target'] = le.fit_transform(obe_cleaned['NObeyesdad'])

# Print the mapping dictionary for your report reference
obesity_mapping = dict(zip(le.classes_, le.transform(le.classes_)))
print("Numeric mapping for Obesity Levels:\n", obesity_mapping)

# 2. Separate features (X) and the target variable (y)
X = obe_cleaned.drop(columns=['NObeyesdad', 'Target'])
y = obe_cleaned['Target']

# 3. Apply One-Hot Encoding to categorical features (e.g., Gender, SMOKE)
# drop_first=True prevents the dummy variable trap (multicollinearity)
X_encoded = pd.get_dummies(X, drop_first=True)

print("\nFirst two rows of the encoded features (X_encoded):")
display(X_encoded.head(2))
```

```python
# ============================================================
# CLEAN OUTLIERS USING IQR CAPPING
# Age, Height, Weight
# ============================================================

import pandas as pd

features = ['Age', 'Height', 'Weight']

for column in features:

    # Calculate Q1 and Q3
    Q1 = obe_cleaned[column].quantile(0.25)
    Q3 = obe_cleaned[column].quantile(0.75)

    # Calculate IQR
    IQR = Q3 - Q1

    # Define lower and upper bounds
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Count outliers before cleaning
    outlier_count = (
        (obe_cleaned[column] < lower_bound) |
        (obe_cleaned[column] > upper_bound)
    ).sum()

    print(f"\n{column}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Outliers Before Cleaning: {outlier_count}")

    # --------------------------------------------------------
    # CAPPING METHOD
    # Values below lower bound become lower bound
    # Values above upper bound become upper bound
    # --------------------------------------------------------

    obe_cleaned[column] = obe_cleaned[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )


print("\nOutlier capping completed.")
```

```python
# ============================================================
# 7. CONFUSION MATRIX EVALUATION FUNCTION
# ============================================================

def evaluate_from_confusion_matrix(
    model_name,
    y_true,
    y_pred
):

    # ---------------------------------------------
    # Create multiclass confusion matrix
    # ---------------------------------------------

    class_labels = np.unique(y_true)

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=class_labels
    )


    total_samples = cm.sum()

    class_results = []


    # ---------------------------------------------
    # Calculate each class separately
    # ---------------------------------------------

    for i, class_name in enumerate(
        class_labels
    ):

        # =========================================
        # TRUE POSITIVE
        # =========================================

        TP = cm[i, i]


        # =========================================
        # FALSE NEGATIVE
        # =========================================
        #
        # Actual class = current class
        # Predicted = another class
        #

        FN = (
            cm[i, :].sum()
            - TP
        )


        # =========================================
        # FALSE POSITIVE
        # =========================================
        #
        # Actual = another class
        # Predicted = current class
        #

        FP = (
            cm[:, i].sum()
            - TP
        )


        # =========================================
        # TRUE NEGATIVE
        # =========================================

        TN = (
            total_samples
            - TP
            - FP
            - FN
        )


        # =========================================
        # PRECISION
        # =========================================

        if (TP + FP) > 0:

            precision = (
                TP /
                (TP + FP)
            )

        else:

            precision = 0


        # =========================================
        # RECALL / SENSITIVITY
        # =========================================

        if (TP + FN) > 0:

            recall = (
                TP /
                (TP + FN)
            )

        else:

            recall = 0


        # =========================================
        # SPECIFICITY
        # =========================================

        if (TN + FP) > 0:

            specificity = (
                TN /
                (TN + FP)
            )

        else:

            specificity = 0


        # =========================================
        # ONE-VS-REST ACCURACY
        # =========================================

        accuracy_class = (
            (TP + TN) /
            (
                TP
                + TN
                + FP
                + FN
            )
        )


        # =========================================
        # F1-SCORE
        # =========================================

        if (
            precision + recall
        ) > 0:

            f1 = (
                2
                * precision
                * recall
                /
                (
                    precision
                    + recall
                )
            )

        else:

            f1 = 0


        # =========================================
        # SAVE CLASS RESULT
        # =========================================

        class_results.append({

            'Class': class_name,

            'TP': TP,
            'TN': TN,
            'FP': FP,
            'FN': FN,

            'Accuracy': accuracy_class,

            'Precision': precision,

            'Recall': recall,

            'Specificity': specificity,

            'F1-Score': f1
        })


    # Convert into dataframe
    class_df = pd.DataFrame(
        class_results
    )


    # ========================================================
    # OVERALL MULTICLASS ACCURACY
    # =======================================================

    overall_accuracy = (
        np.trace(cm)
        /
        cm.sum()
    )


    # ========================================================
    # MACRO AVERAGE
    # ========================================================

    macro_precision = (
        class_df[
            'Precision'
        ].mean()
    )


    macro_recall = (
        class_df[
            'Recall'
        ].mean()
    )


    macro_specificity = (
        class_df[
            'Specificity'
        ].mean()
    )


    macro_f1 = (
        class_df[
            'F1-Score'
        ].mean()
    )


    overall_result = {

        'Model': model_name,

        'Accuracy':
            overall_accuracy,

        'Macro Precision':
            macro_precision,

        'Macro Recall':
            macro_recall,

        'Macro Specificity':
            macro_specificity,

        'Macro F1-Score':
            macro_f1
    }


    return (
        overall_result,
        class_df,
        cm,
        class_labels
    )
```

```python

from sklearn.metrics import confusion_matrix

# ============================================================
# 1. FUNCTION TO CALCULATE PER-CLASS METRICS
# ============================================================

def get_per_class_confusion_metrics(
    model_name,
    y_true,
    y_pred
):

    # Get all classes
    class_labels = np.unique(y_true)

    # Create multiclass confusion matrix
    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=class_labels
    )

    total_samples = cm.sum()

    rows = []


    # ========================================================
    # LOOP THROUGH EACH CLASS
    # ========================================================

    for i, class_name in enumerate(class_labels):

        TP = cm[i, i]

        FN = (
            cm[i, :].sum()
            - TP
        )

        FP = (
            cm[:, i].sum()
            - TP
        )

        TN = (
            total_samples
            - TP
            - FP
            - FN
        )


        # ====================================================
        # ACCURACY
        # ====================================================

        accuracy = (
            (TP + TN)
            /
            (TP + TN + FP + FN)
        )


        # ====================================================
        # PRECISION
        # ====================================================

        if (TP + FP) > 0:

            precision = (
                TP
                /
                (TP + FP)
            )

        else:

            precision = 0


        # ====================================================
        # RECALL
        # ====================================================

        if (TP + FN) > 0:

            recall = (
                TP
                /
                (TP + FN)
            )

        else:

            recall = 0


        # ====================================================
        # F1-SCORE
        # ====================================================

        if (precision + recall) > 0:

            f1 = (
                2
                * precision
                * recall
                /
                (precision + recall)
            )

        else:

            f1 = 0


        # ====================================================
        # ADD ONE ROW
        # ====================================================

        rows.append({

            'Model': model_name,

            'Class': class_name,

            'TP': TP,

            'TN': TN,

            'FP': FP,

            'FN': FN,

            'Accuracy': accuracy,

            'Precision': precision,

            'Recall': recall,

            'F1-Score': f1
        })


    return pd.DataFrame(rows)


# ============================================================
# 2. DECISION TREE
# ============================================================

dt_details = get_per_class_confusion_metrics(
    'Decision Tree',
    y_test,
    dt_pred
)


# ============================================================
# 3. NAIVE BAYES
# ============================================================

nb_details = get_per_class_confusion_metrics(
    'Naive Bayes',
    y_test,
    nb_pred
)


# ============================================================
# 4. LOGISTIC REGRESSION
# ============================================================

lr_details = get_per_class_confusion_metrics(
    'Logistic Regression',
    y_test,
    lr_pred
)


# ============================================================
# 5. COMBINE ALL MODELS
# ============================================================

all_21_results = pd.concat(
    [
        dt_details,
        nb_details,
        lr_details
    ],
    ignore_index=True
)


# ============================================================
# 6. DISPLAY AS DECIMAL
# ============================================================

display(
    all_21_results.round(4)
)

```

```python
# ============================================================
# FIND THE BEST MODEL
# Based on Macro F1-Score
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

import pandas as pd


# ============================================================
# 1. EVALUATION FUNCTION
# ============================================================

def evaluate_model(model_name, y_true, y_pred):

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average='macro',
        zero_division=0
    )

    return {
        'Model': model_name,
        'Accuracy': accuracy,
        'Macro Precision': precision,
        'Macro Recall': recall,
        'Macro F1-Score': f1
    }


# ============================================================
# 2. CALCULATE RESULTS FOR ALL MODELS
# ============================================================

dt_result = evaluate_model(
    'Decision Tree',
    y_test,
    dt_pred
)

nb_result = evaluate_model(
    'Naive Bayes',
    y_test,
    nb_pred
)

lr_result = evaluate_model(
    'Logistic Regression',
    y_test,
    lr_pred
)


# ============================================================
# 3. CREATE RESULT TABLE
# ============================================================

results_df = pd.DataFrame([
    dt_result,
    nb_result,
    lr_result
])


# ============================================================
# 4. DISPLAY AS PERCENTAGE
# ============================================================

results_display = results_df.copy()

metric_columns = [
    'Accuracy',
    'Macro Precision',
    'Macro Recall',
    'Macro F1-Score'
]

for col in metric_columns:

    results_display[col] = (
        results_display[col] * 100
    ).round(2).astype(str) + '%'


print("=" * 90)
print("MODEL PERFORMANCE COMPARISON")
print("=" * 90)

display(results_display)


# ============================================================
# 5. FIND BEST MODEL
# ============================================================

best_index = results_df[
    'Macro F1-Score'
].idxmax()

best_model_name = results_df.loc[
    best_index,
    'Model'
]

best_accuracy = results_df.loc[
    best_index,
    'Accuracy'
]

best_precision = results_df.loc[
    best_index,
    'Macro Precision'
]

best_recall = results_df.loc[
    best_index,
    'Macro Recall'
]

best_f1 = results_df.loc[
    best_index,
    'Macro F1-Score'
]


print("\n")
print("=" * 70)
print("BEST MACHINE LEARNING MODEL")
print("=" * 70)

print(
    f"Best Model      : {best_model_name}"
)

print(
    f"Accuracy        : {best_accuracy * 100:.2f}%"
)

print(
    f"Macro Precision : {best_precision * 100:.2f}%"
)

print(
    f"Macro Recall    : {best_recall * 100:.2f}%"
)

print(
    f"Macro F1-Score  : {best_f1 * 100:.2f}%"
)
```
